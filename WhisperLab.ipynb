## Step 3: Simplify the Script for Colab

```python
import whisper, json, os
from datetime import datetime
from pathlib import Path
import warnings
warnings.filterwarnings("ignore", category=FutureWarning)

class AudioTranscriber:
    def __init__(self, model_size="large"):
        print(f"Loading Whisper model: {model_size}...")
        self.model = whisper.load_model(model_size)
        print(f"✓ Model {model_size} loaded successfully")

    def transcribe_audio(self, audio_path, language=None, task="transcribe"):
        print(f"🎧 Transcribing: {audio_path}")
        options = {"task": task, "fp16": False}
        if language:
            options["language"] = language
        result = self.model.transcribe(audio_path, **options)
        data = {
            "file": os.path.basename(audio_path),
            "text": result["text"].strip(),
            "language": result.get("language", "unknown"),
            "segments": result.get("segments", []),
            "date": datetime.now().isoformat()
        }
        return data

    def save_transcription(self, data, filename="transcription.json"):
        with open(filename, "w", encoding="utf-8") as f:
            json.dump(data, f, indent=2, ensure_ascii=False)
        print(f"💾 Saved: {filename}")
```

## Step 4: Run the Transcription

```python
transcriber = AudioTranscriber(model_size="large")  # or "medium" / "large"
result = transcriber.transcribe_audio(audio_path)
transcriber.save_transcription(result, "output.json")

print("\n📝 Transcribed Text:\n")
print(result["text"])
```
